# 2. Data Understanding

## Dataset Description

The dataset contains detailed information about Sephora beauty products and customer reviews. It includes product information such as brand, category, price, rating, loves count, ingredients, and customer feedback.

## Column Definitions

### Product Information Dataset

- **product_id** – Unique identifier for each product.
- **product_name** – Name of the product.
- **brand_name** – Brand of the product.
- **primary_category** – Main category of the product.
- **secondary_category** – Sub-category.
- **tertiary_category** – Detailed category.
- **price_usd** – Selling price in USD.
- **value_price_usd** – Original value price.
- **rating** – Average customer rating.
- **loves_count** – Number of users who marked the product as favorite.
- **ingredients** – Ingredients used in the product.
- **highlights** – Key product features.
- **online_only** – Product available only online.
- **out_of_stock** – Product stock status.
- **sephora_exclusive** – Indicates whether the product is exclusive to Sephora.

### Reviews Dataset

- **review_id** – Unique review identifier.
- **author_id** – Unique customer identifier.
- **product_id** – Product identifier linked to the review.
- **rating** – Rating given by the customer.
- **review_text** – Customer review or comment.
- **submission_time** – Date and time of the review.
- **is_recommended** – Whether the customer recommends the product.
- **helpfulness** – Number of users who found the review helpful.

# Key Findings and Recommendations

## Key Findings

- Most Sephora products have ratings between 4.0 and 5.0, indicating high customer satisfaction.
- Product prices vary significantly, with most products in the low to medium price range.
- Loves Count is highly skewed, with only a few products receiving very high customer engagement.
- Product price has a weak relationship with customer ratings.
- Ratings and Loves Count also show only a weak positive correlation.

## Recommendations

- Promote products with high ratings and high Loves Count.
- Improve products with lower ratings by considering customer feedback.
- Increase marketing efforts for high-performing product categories.
- Encourage customers to leave more reviews to improve engagement.
